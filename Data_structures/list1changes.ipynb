```python
# Real Life Example (Very Easy to Remember)
# Shallow Copy → Google Docs Sharing

# Two people open the same Google Doc.

# If person A edits the document →
# Person B immediately sees the change.

# Because both are using the same document.
```

```python
import copy

list1 = [1, 2, [3, 4]]

list2 = copy.copy(list1)

print("Outer list memory:")
print(list1)
print(id(list1), id(list2))

print("Inner list memory:")
print(list2)
print(id(list1[2]), id(list2[2]))
```

```python
#     Deep Copy → Downloaded File

# You download a copy of a document.

# Now you edit your file.

# The original file does not change.

# Because it is a separate copy.
```

```python
import copy

list1 = [1, 2, [3, 4]]

list2 = copy.deepcopy(list1)

print("Outer list memory:")
print(id(list1), id(list2))

print("Inner list memory:")
print(id(list1[2]), id(list2[2]))
```
