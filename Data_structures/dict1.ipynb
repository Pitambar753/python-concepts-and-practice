# Dictionary
Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items->dict,list,set

1. 
-  Dictionary keys are case sensitive: the same name but different cases of Key will be treated distinctly.
- Keys must be immutable: This means keys can be strings, numbers or tuples but not lists.
- Keys must be unique: Duplicate keys are not allowed and any duplicate key will overwrite the previous value.
- Dictionary internally uses Hashing. Hence, operations like search, insert, delete can be performed in Constant Time
- dictionary is unordered 

```python
# empty dictionary
d = {}
d
# 1D dictionary
d1 = { 'name' : 'nitish' ,'gender' : 'male' }
print(d1)
# with mixed keys
d2 = {(1,2,3):1,'hello':'world'}
print(d2)
# 2D dictionary -> JSON
s = {
    'name':'nitish',
     'college':'bit',
     'sem':4,
     'subjects':{
         'dsa':50,
         'maths':67,
         'english':34
     }
}
print(s)
# using sequence and dict function
d4 = dict([('name','nitish'),('age',32),(3,3)])
print("d4->",d4)
# duplicate keys
d5 = {'name':'nitish','name':'rahul'}
print("d5->",d5)
# mutable items as keys
d6 = {'name':'nitish',(1,2,3):2}
print("d6->",d6)
s['subjects']['dsa'] = 80
s
```

```python
d = {1: 'Geeks', 2: 'For', 3: 'Geeks', 'age':22}

# Using del to remove an item
del d["age"]
print(d)

# Using pop() to remove an item and return the value
val = d.pop(1)
print(val)

# Using popitem to removes and returns
# the last key-value pair.
key, val = d.popitem()
print(f"Key: {key}, Value: {val}")

# Clear all items from the dictionary
d.clear()
print(d)
```

```python
# get()
# key (required): The key you want to look up.
# default_value (optional): Value to return if the key is not found. Defaults to None.

# Why use get() instead of dict[key]?
# Using dict[key] raises a KeyError if the key is missing.
# Using dict.get(key) returns None or a default value, making it safer for optional lookups.
```

```python
# Example dictionary
student = {
    "name": "Alice",
    "age": 21,
    "course": "Computer Science"
}

# 1. Get existing key
print(student.get("name"))  
# Output: Alice

# 2. Get non-existing key (returns None by default)
print(student.get("grade"))  
# Output: None

# 3. Get non-existing key with a default value
print(student.get("grade", "Not Assigned"))  
# Output: Not Assigned
```
