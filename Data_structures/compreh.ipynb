## zip fuction 


- The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

- If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# Write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

list(zip(L1,L2)) 

[i+j for i,j in zip(L1,L2)]
```

## list comprehension 1



### Nested list comprehension — 2D matrix

```python
### Two for loops inside — creates a flat list from a 2D structure. Common CCEE output prediction question.
matrix = [[1,2,3], [4,5,6], [7,8,9]]
flat = [n for row in matrix for n in row]
print(flat)

# Only elements > 5
big = [n for row in matrix for n in row if n > 5]
print(big)
```


### if-else inside comprehension (ternary)
- if-else in the EXPRESSION part (before for) vs if at the END (filter) — very different!
- This is one of the MOST common CCEE MCQs! if-else before for = transform all. if after for = filter some.

```python
nums = [1, 2, 3, 4, 5]

# if-else in expression (ALL items kept, value changes)
r1 = ["E" if x%2==0 else "O" for x in nums]
print(r1)

# if at end (filters, only SOME items kept)
r2 = [x for x in nums if x%2==0]
print(r2)
```

###  Dict & Set comprehension

```python
# Dict comprehension
squares_dict = {x: x**2 for x in range(1, 6)}
print(squares_dict)

# Set comprehension (no duplicates)
data = [1, 2, 2, 3, 3, 3]
unique_sq = {x**2 for x in data}
print(unique_sq)
```

### List comprehension vs map vs loop — speed
- All 3 do the same thing. List comprehension is fastest and most Pythonic for CDAC.

```python
# All 3 produce same result
data = range(10)

# 1. for loop
r1 = []
for x in data: r1.append(x**2)

# 2. map()
r2 = list(map(lambda x: x**2, data))

# 3. List comprehension (preferred)
r3 = [x**2 for x in data]
print(r1 == r2 == r3)
```
