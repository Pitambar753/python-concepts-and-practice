```python
print(pow(2, 3))  # Output: 8 (2 raised to the power of 3)
print(pow(5, 2))  # Output: 25 (5 squared)
print(pow(2, -2))  # Output: 0.25 (2 raised to the power of -2 → 1/(2²))
print(pow(3.5, 2))  # Output: 12.25 (3.5 squared)
```

```python
print(pow(2, 3, 5))  # Output: 3 (2³ = 8, and 8 % 5 = 3)
print(pow(7, 2, 4))  # Output: 1 (7² = 49, and 49 % 4 = 1)
```
