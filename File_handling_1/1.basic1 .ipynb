## Using Context Manager (With)

- It's a good idea to close a file after usage as it will free up the resources
- If we dont close it, garbage collector would close it
- ***with*** keyword closes the file as soon as the usage is over

```python
with open('text/sample.txt','w') as f:
    
    f.write("salman bhai ki jay")
```

```python
f.write("helloe ")


# jar apan **with** keyword use kela tar automatically file close hoti 
```

```python
# try f.read() now
with open("text/sample.txt", 'r') as f:
    # print(f.read())
    print(f.readline())
```

## seek and tell function
 - tell --> then you can tell the buffer where it is right now , tell --> tell you to where are you right now with indexing
 - seek --> go back to the give position , seek -->  work for you can go anywhere add the index no only  

```python
with open('text/list1.txt','r') as f:
    print(f.read(10))
    print(f.tell())
    f.seek(0)     
    print(f.tell())   
    print(f.read(10))
    print(f.read(10))
```

```python
# seek during write
with open('text/sample.txt','w') as f:
  f.write('Hello')
  f.seek(0)     # seek like youtube video red line ke jasa hai ti ahe karan kuthe pan jataya
  f.write('X')
```

## Problems with working in text mode

- can't work with binary files like images
- not good for other data types like int/float/list/tuples

```python
# working with binary file using text mode
with open('text/Screenshot (1).png','r') as f:
  f.read()
```

```python
# working with binary file using binary mode
with open('text/Screenshot (1).png','rb') as f:
    with open('text/Screenshot2.png','wb') as wf:
        wf.write(f.read())
```

```python
# working with other data types
with open('text/sample.txt','w') as f:
  f.write(5)  ### beacuse we not add another type only add the string for unicode
```

```python
# working with other data types
with open('text/sample.txt','w') as f:
  f.write('5')
```

```python
with open('text/sample.txt','r') as f:
  print(int(f.read()) + 5)
```

```python
# more complex data
d = {
    'name':'nitish',
     'age':33,
     'gender':'male'
}

with open('text/sample.txt','w') as f:
  f.write(str(d))
```
