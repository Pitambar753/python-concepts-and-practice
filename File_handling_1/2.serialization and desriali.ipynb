### Serialization and Deserialization

#### - **Serialization** - process of converting python data types to JSON format
- Serialization is the process of converting a Python object (like a dictionary, list, class object, etc.) into a format that can be stored (e.g., in a file or database) or transmitted (e.g., over a network).
Common formats: JSON, pickle, XML
#### - **Deserialization** - process of converting JSON to python data types
- Deserialization is the reverse process — it converts the stored format back into the original Python object.

1. JSON works with standard data types like dict, list, string, int, float, etc.

```python
# serialization using json module
# list
import json
 
L = [1,2,3,4]

with open('text/demo.json','w') as f:
    json.dump(L,f)
```

```python
#### # deserialization

with open('text/demo.json','r') as f:
    print(json.load(f))
```

```python
# dict
d = {
    'name':'nitish',
     'age':33,
     'gender':'male'
}

with open('text/demo.json','w') as f:
    json.dump(d,f,indent=4)
```

```python
# deserialization
import json

with open('text/demo.json','r') as f:
    d = json.load(f)
    print(d)
    print(type(d))
```

```python
# serialize and deserialize with tuple
import json

t = (1,2,3,4,5)   # tuple send kel ki list madhe convert hotya 

with open('text/demo1.json','w') as f:
  json.dump(t,f)

```

## Serializing and Deserializing custom objects

```python
class Person:

  def __init__(self,fname,lname,age,gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

# format to printed in
# -> piti salgatre age -> 22 gender -> male
```

```python
person = Person('pitambar','salgare',22,'male')
```

#### 1. json does not directly support custom objects, so we must define how to encode and decode using __dict__

```python

import json
with open ("text/object.json", 'w') as f1:
    json.dump(person,f1)
# python ke readymate data type ko serialsize karu shoto pan 
# but you want serilize the own class ke custom object ko serialsize kar nahi kara shakate
```

```python
import json 
def show_object (person):
    if isinstance(person,Person):
        return "{} {} age --> {} gender -->{}".format(person.fname, person.lname,person.age,person.gender)

with open ('demo.json','w') as f:
    json.dump(person,f, default =show_object)
```

```python
import json

# Define a class
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Create object
s1 = Student("Pitambar", 21)

# ✅ Serialize using custom function
def encode_student(obj):
    if isinstance(obj, Student):
        return obj.__dict__  # Convert object to dictionary
    raise TypeError("Object of type Student is not JSON serializable")

json_data = json.dumps(s1, default=encode_student)
print("Serialized JSON:", json_data)



# ✅ Deserialize using custom function
def decode_student(dct):
    return Student(dct['name'], dct['age'])

loaded_student = json.loads(json_data, object_hook=decode_student)
print("Deserialized:", loaded_student.name, loaded_student.age)
```

### Pickling
- `Pickling` is the process whereby a Python object hierarchy is converted into a byte stream, and 
- `unpickling` is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

```python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
```

```python
p = Person('pitambar',23)
```

```python
# pickle dump
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
  p = pickle.load(f)

p.display_info()
```

### Pickle Vs Json

- Pickle lets the user to store data in binary format.
- JSON lets the user store data in a human-readable text format.
