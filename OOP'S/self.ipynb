# self in Python class
 - It represents the instance of the class being used.
 -  Whenever we create an object from a class, self refers to the current object instance.
 -   It is essential for accessing attributes and methods within the class.

```python
# Define a class called Mynumber to represent a number object
class Mynumber:
    # Constructor method that initializes a new Mynumber object with a value
    def __init__(self, value):
        self.value = value
    
    # Method to print the stored value to the console
    def print_value(self):
        print(self.value)

# Create an instance of Mynumber with value 17
obj1 = Mynumber(17)     # (sel.value == obj1.value)  same ase ahe 
# Call the print_value method to display the value (17)

obj2 = Mynumber(45)
print(obj1.value)
print(obj2.value)   #  
```

```python
class Subject:

    def __init__(self, attr1, attr2):
        self.attr1 = attr1
        self.attr2 = attr2


obj = Subject('Maths', 'Science')
print(obj.attr1)  
print(obj.attr2) 

```

```python
class Car:
    def __init__(self, model, color):
        self.model = model
        self.color = color

    def show(self):
        print("Model is", self.model)
        print("Color is", self.color)

# Creating instances of the class
audi = Car("Audi A4", "Blue")
ferrari = Car("Ferrari 488", "Green")

# Calling instance methods
audi.show()  
ferrari.show()  
```

```python
#Although self is not a Python keyword, using it is a widely accepted convention.
class Car:
    def __init__(this, model, color):
        this.model = model
        this.color = color

    def show(this):
        print("Model is", this.model)
        print("Color is", this.color)

# using 'this' instead of 'self'
audi = Car("Audi A4", "Blue")
audi.show()
```

1. self refers to the memory address of the object and we can use id(self) to see its unique memory address

```python
class Check:
    def __init__(self):
        print("Address of self = ", id(self))

obj = Check()
print("Address of class object = ", id(obj))
```

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

    def decrement(self):
        self.count -= 1

    def get_count(self):
        return self.count

# using the Counter class
counter1 = Counter()
counter1.increment()
counter1.increment()
counter1.decrement()
print(counter1.get_count())
```

```python
# Write Python code here
class sampleclass:
    count = 0     # class attribute

    def increase(self):
        sampleclass.count += 1

# Calling increase() on an object
s1 = sampleclass()
s1.increase()        
print(s1.count)

# Calling increase on one more
# object
s2 = sampleclass()
s2.increase()
print(s2.count)

print(sampleclass.count)
# print(dir(s2))
```

```python
class Student:
    school = "ABC High"  # Class variable (shared by all instances)

    def __init__(self, name):
        self.name = name  # Instance variable

# Class variable is shared
print(Student.school)  # Output: "ABC High"
student1 = Student("Alice")
print(student1.school)  # Output: "ABC High"
```

```python
class Circle:
    def __init__(self, r):
        self.r = r

    def area(self):
        a = 3.14 * self.r ** 2
        return a

# Creating an instance of Circle
ins = Circle(5)

# Calling the area method
print("Area of the circle:", ins.area())
```
