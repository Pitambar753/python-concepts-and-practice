### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
# object without a reference
class Person:

  def __init__(self):
    self.name = 'nitish'
    self.gender = 'male'

p = Person()
q = p
```

```python
# Multiple ref
print(id(p))
print(id(q))
print(id(Person()))
print(Person)
```

```python
print(p.name)
print(q.name)
q.name =  "pitu"
print(p.name)
print(q.name)
```

# Object ki mutability

```python
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  person.name = 'ankit'
  return person

p = Person('nitish','male')
print(id(p))
p1 = greet(p)
print(id(p1))
```

```python
class Customer:
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender
        
def greet(Customer1):
    if "male" == Customer1.gender:
        print("hii",Customer1.name,"sir")
    else :
        print("hii",Customer1.name,"ma'ma")
    
    cust2 = Customer("anikita","female")
    return cust2

cust = Customer("pitu","male")
new_cust = greet(cust)
print(new_cust.name)
```

```python
# Part: Professor class
class Professor:
    def __init__(self, name):
        self.name = name

# Whole: University class
class University:
    def __init__(self, name):
        self.name = name
        self.professors = []  # List to store professors

    def add_professor(self, professor):
        self.professors.append(professor)

# Creating objects
prof1 = Professor("Dr. Smith")
prof2 = Professor("Dr. Jones")

# Creating a university and adding professors
uni = University("ABC University")
uni.add_professor(prof1)
uni.add_professor(prof2)

# Displaying professors
for prof in uni.professors:
    print(f"{prof.name} works at {uni.name}")
    
print(uni.professors[0].name)  # Output: Dr. Smith
print(uni.professors[1].name)  # Output: Dr. Jones
```
