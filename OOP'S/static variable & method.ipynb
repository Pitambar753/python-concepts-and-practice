```python
class Student:
    school = "ABC School"  # Static (class) variable

    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age    # Instance variable

# Access static variable (no object needed)
print(Student.school)  # Output: "ABC School"

# Create objects
student1 = Student("Alice", 20)
student2 = Student("Bob", 22)

# All objects share the same static variable
print(student1.school)  # Output: "ABC School"
print(student2.school)  # Output: "ABC School"

# Instance variables are unique per object
print(student1.name)  # Output: "Alice"
print(student2.name)  # Output: "Bob"

# Modify static variable (affects all objects)
Student.school = "XYZ School"
print(student1.school)  # Output: "XYZ School" (updated)
```

```python
class Employee:
    company = "Tech Corp"  # Static variable (shared by all employees)

    def __init__(self, name, salary):
        self.name = name    # Instance variable (unique per employee)
        self.salary = salary  # Instance variable

# Access static variable
print(Employee.company)  # Output: "Tech Corp"

# Create objects
emp1 = Employee("Alice", 50000)
emp2 = Employee("Bob", 60000)

# All objects share the same static variable
print(emp1.company)  # Output: "Tech Corp"
print(emp2.company)  # Output: "Tech Corp"

# Instance variables are unique per object
print(emp1.name)  # Output: "Alice"
print(emp2.name)  # Output: "Bob"

# Modify static variable (changes reflect for all objects)
Employee.company = "New Tech Corp"
print(emp1.company)  # Output: "New Tech Corp"
```

## Static methods
##### Points to remember about static

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
- The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

```python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,"drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b

# Call without creating an object
print(MathUtils.add(5, 3))        # Output: 8
print(MathUtils.multiply(5, 3))   # Output: 15
```


### CCEE Trap Line (Memorize This)
- Mutable class variables are shared, but assignment creates instance variables.

```python
class A:
    x = 10

a = A()
b = A()

print(a.x, b.x)
print(a.__dict__)
print(A.__dict__)
```

```python
class A:
    x = 10

a = A()
b = A()

A.x = 20

print(a.x, b.x)
```

```python
class A:
    x = 10

a = A()
b = A()

a.x = 50

print(a.x, b.x)
```

```python
# ✔ Same list shared across all instances
class A:
    lst = []

a = A()
b = A()

a.lst.append(1)

print(b.lst)
```

```python

# class list unchanged
class A:
    lst = []

a = A()
b = A()

a.lst = [5]

print(b.lst)
print(a.lst)    # a.lst = [5] → NEW instance variable
```

```python
# ✔ First mutation affects all
# ✔ Reassignment affects only a

class A:
    lst = []

a = A()
b = A()

a.lst.append(1)
a.lst = [2]

print(b.lst)
print(a.lst)
```
