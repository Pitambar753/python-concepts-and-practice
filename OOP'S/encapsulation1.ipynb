### Data Hiding: 
 - The variables (attributes) are kept private or protected,
 - meaning they are not accessible directly from outside the class. 
 - Instead, they can only be accessed or modified through the methods.

    
    
# Access through Methods: 
 - Methods act as the interface through which external code interacts with the data stored in the variables. 
For instance, getters and setters are common methods used to retrieve and update the value of a private variable.




# Control and Security: 
 - By encapsulating the variables and only allowing their manipulation via methods, 
the class can enforce rules on how the variables are accessed or modified,
thus maintaining control and security over the data

```python
          

# Encapsulation is the process of hiding the internal state of an 
# object and requiring all interactions to be performed through an object’s methods. This approach:

# Provides better control over data.
# Prevents accidental modification of data.
# Promotes modular programming.
```

```python
# Encapsulation in Python is like having a bank account system where your account balance (data) is kept private. 
# You can’t directly change your balance by accessing the account database. 
# Instead, the bank provides you with methods (functions) like deposit and withdraw to modify your balance safely.

# Private Data (Balance): 
# Your balance is stored securely. Direct access from outside is not allowed, ensuring the data is protected from unauthorized changes.


# Public Methods (Deposit and Withdraw): These are the only ways to modify your balance.
# They check if your requests (like withdrawing money) follow the rules (e.g., you have enough balance) before allowing changes.
```

```python
# Public members are accessible from anywhere, both inside and outside the class. 
# These are the default members in Python.

    
class Public:
    def __init__(self):
        self.name = "John"  # Public attribute

    def display_name(self):   # Public method
        print(self.name)  

obj = Public()
obj.display_name()  # Accessible
print(obj.name)  # Accessible
```

1. Public Attribute (name): 
 - This attribute is declared without any underscore prefixes. It is accessible from anywhere, both inside and outside of the class.


2. Public Method (display_name): 
 - This method is also accessible from any part of the code. It directly accesses the public attribute and prints its value.

    
3. Object (obj): 
 - An instance of Public is created, and the display_name method is called, demonstrating how public attributes and methods can be accessed directly

# Protected members
 - Protected members are identified with a single underscore (_). 
        They are meant to be accessed only within the class or its subclasses

```python


class Protected:
    def __init__(self):
        self._age = 30  # Protected attribute

class Subclass(Protected):
    def display_age(self):
        print(self._age)  # Accessible in subclass

obj = Subclass()
obj.display_age()
obj1 = Protected()
print(obj._age)    # 
```

# Private members
- Private members are identified with a double underscore (__) and cannot be accessed directly from outside the class.
- but accessed in inside the class
- and also you can access using  (_class_name__variable_name) 
Python uses name mangling to make private members inaccessible by renaming them internally

```python
class Private:
    def __init__(self):
        self.__salary = 50000  # Private attribute

    def salary_1(self):
        return self.__salary  # Access through public method

obj = Private()
print(obj.salary_1())  # Works
#print(obj.__salary)  # Raises AttributeError
```

```python
# program to illustrate public access modifier in a class


class Geek:

    # constructor
    def __init__(self, name, age):

        # public data members
        self.geekName = name
        self.geekAge = age

    # public member function
    def displayAge(self):

        # accessing public data member
        print("Age: ", self.geekAge)


# creating object of the class
obj = Geek("R2J", 20)

# finding all the fields and methods which are present inside obj
print("List of fields and methods inside obj:", dir(obj))

# accessing public data member
print("Name:", obj.geekName)

# calling public member function of the class
obj.displayAge()
```

```python
#Public members (brand, speed) can be accessed anywhere.
class Car:
    def __init__(self, brand, speed):
        self.brand = brand  # Public variable
        self.speed = speed  # Public variable

    def show_details(self):
        print(f"Car Brand: {self.brand}, Speed: {self.speed} km/h")

# Creating an object
car1 = Car("Tesla", 120)

# Accessing public members directly
print(car1.brand)  # ✅ Allowed
print(car1.speed)  # ✅ Allowed

# Calling a public method
car1.show_details()  # ✅ Allowed
```

```python
# Protected members (_name, _species) should be accessed only within the class or subclass.
class Animal:
    def __init__(self, name, species):
        self._name = name  # Protected variable
        self._species = species  # Protected variable

    def _display_info(self):  # Protected method
        print(f"Animal: {self._name}, Species: {self._species}")

class Dog(Animal):
    def show(self):
        print(f"Dog's Name: {self._name}")  # Accessing protected variable
        self._display_info()  # Accessing protected method

# Creating an object
dog1 = Dog("Buddy", "Golden Retriever")

# Accessing protected members (Not recommended but possible)
print(dog1._name)  # ✅ Allowed but should be avoided

# Accessing protected method (Not recommended but possible)
dog1._display_info()  # ✅ Allowed but should be avoided

# Using subclass method
dog1.show()  # ✅ Allowed
```

```python

class Employee:
    def __init__(self, name, salary):
        self.name = name  # Public variable
        self.__salary = salary  # Private variable

    def __calculate_bonus(self):  # Private method
        return self.__salary * 0.10  # Bonus is 10% of salary

    def get_bonus(self):
        # Accessing the private method inside the class
        return f"Bonus for {self.name}: {self.__calculate_bonus()}"

# Creating an object
emp1 = Employee("John", 50000)

# Trying to access private method directly (Will  an error)
# print(emp1.__calculate_bonus())   Error

# Accessing private method using a public method
print(emp1.get_bonus())  #  Allowed
```

```python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self._account_holder = account_holder  # Protected
        self.__balance = balance              # Private (mangled to _BankAccount__balance)

    # Getter for balance (read-only)
    def get_balance(self):
        return self.__balance

    # Setter for controlled updates
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid amount!")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.__balance}")
        else:
            print("Insufficient funds!")

# Usage
account = BankAccount("Alice", 1000)
print(account._account_holder)       # Works (protected)
# print(account.__balance)           # Error! AttributeError (private)

account.deposit(500)                 # Output: Deposited $500. New balance: $1500
account.withdraw(2000)               # Output: Insufficient funds!
print(account.get_balance())         # Output: 1500 (via getter)
```

```python
class Car:
    def __init__(self):
        self.__max_speed = 120  # Private
        self._current_speed = 0  # Protected
        self.speed = 50
    def accelerate(self, increase):
        new_speed = self._current_speed + increase
        if new_speed <= self.__max_speed:
            self._current_speed = new_speed
        else:
            self._current_speed = self.__max_speed
            print("Max speed reached!")

car = Car()
car.accelerate(100)
print(car._current_speed)  # Output: 100
car.accelerate(30)  # Output: Max speed reached!
print(car._current_speed)
print(car.speed)
print(car.__max_speed)   # private jar acess karayach try jar kele tar nahi milanar
 # car.
```
