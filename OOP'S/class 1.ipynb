```python
class Dog:
    def __init__(self, name, breed="Mixed", age=1):
        self.name = name
        self.breed = breed
        self.age = age

# Creating an instance with default parameters
dog1 = Dog("Buddy")
print(dog1.name)   # Output: Buddy
print(dog1.breed)  # Output: Mixed
print(dog1.age)    # Output: 1

# Creating an instance with specified parameters
dog2 = Dog("Max", "Golden Retriever", 5)
print(dog2.name)   # Output: Max
print(dog2.breed)  # Output: Golden Retriever
print(dog2.age)    # Output: 5
```

```python
class Atm:
    def __init__(self):
        self.pin = ""
        self.balances = 0
        self.menu()
  
    
    def menu(self):
        user_input = input(""" 
        hi how can help you 
        1. Presss 1 to create pin 
        2. Press 2 to change pin 
        3. Press 3 to check balances 
        4. Press 4  to withdraw
        5. anything else to exit 
        """)

        if "1" == user_input:
            self.create_pin()
        elif "2" == user_input:
            self.change_pin()
        elif "3" == user_input:
            self.check_balances()
        elif "4" == user_input:
            self.withdraw()
            
        else :
            pass 

    def create_pin(self):
        user_pin = input("enter the pin :")
        self.pin = user_pin

        user_balance = int(input("enter the balances"))
        self.balances = user_balance

        print(" pin created successfully")
        self.menu()

    def change_pin(self):
        old_pin = input("enter the old pin :")
        if self.pin == old_pin:
            New_pin = input(" enter the new pin :")
            self.pin  = New_pin 
            print("pin change successfully")
            self.menu()   # continutie chalu raha mahanuna menu la call kel
        else:
            print("wrong pin takaly tumhi ")
            self.menu()

    def check_balances(self):
        user_pin = input("enter the pin:")
        if user_pin == self.pin :
            print("our current balances is :" ,self.balances )
            self.menu()
        else :
            print("chal jaa be love kar  ")
            self.menu()
            
    def withdraw(self):
        user_pin = input("enter the pin ")
        if user_pin == self.pin:
            amount = int(input("enter the amount "))
            if amount <= self.balances :
                self.balances = self.balances - amount
                print("withdrawl is successfull , balances is ", self.balances )
            else :
                print("abe saale garib")
        else:
            print("chor ahe")
        self.menu()
        
        
sbi = Atm()
```
