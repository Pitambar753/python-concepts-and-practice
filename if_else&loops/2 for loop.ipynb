# For Loop in Python

For loops are used for sequential traversal. For example: traversing a list or string or array etc. 
In Python, there is "for in" loop which is similar to foreach loop in other languages.
Let us learn how to use for loops in Python for sequential traversals with examples.


for var in iterable:

    # statements

    pass


- In Python, for loops only implement the collection-based iteration

### range() Method
range() function is used with for loops to generate a sequence of numbers. It can take one, two or three arguments:

- range(stop) generates numbers from 0 to stop-1.
- range(start, stop) generates numbers from start to stop-1.
- range(start, stop, step) generates numbers from start to stop-1, incrementing by step.

```python
li = ["geeks", "for", "geeks"]
for i in li:
    print(i)
    
tup = ("geeks", "for", "geeks")
for i in tup:
    print(i)
    
s = "Geeks"
for i in s:
    print(i)
    
d = dict({'x':123, 'y':354})
for i in d:
    print("%s  %d" % (i, d[i]))
    
set1 = {1, 2, 3, 4, 5, 6}
for i in set1:
    print(i),
```

```python
for i in range(0, 10, 2):
    print(i)
```

- Continue with For Loop

```python
# Prints all letters except 'e' and 's'

for i in 'geeksforgeeks':

    if i == 'e' or i == 's':
        continue
    print(i)
```

- Break with For Loop

```python
for i in 'geeksforgeeks':

    # break the loop as soon it sees 'e'
    # or 's'
    if i == 'e' or i == 's':
        break

print(i)
```

- Pass Statement with For Loop

```python
# An empty loop
for i in 'geeksforgeeks':
    pass
print(i)
```

```python
for i in range(1, 4):
    print(i)
else:  # Executed because no break in for
    print("No Break\n")
```

# Using Enumerate with for loop
- In Python, enumerate() function is used with the for loop to iterate over an iterable while also keeping track of index of each item.

enumerate(iterable, start=0) 

```python
li = ["eat", "work","apply_For_job", "repeat"]

for i, j in enumerate(li):
    print (i, j)
```

```python
a = ["x", "y", "z"]
e = enumerate(a)

print(next(e))
print(next(e))
```

```python
######### exam0ple 1
a = ["A", "B", "C"]
r = list(enumerate(a))
print(r)
```

```python
d = {"a": 10, "b": 20}
for i, (k, v) in enumerate(d.items()):
    print(i, k, v)
```

```python
a = 6
b = 9
result = max(a, b)
while True:
    if result % a == 0 and result % b == 0:
        break
    result += max(a, b)
    print(result)
```
