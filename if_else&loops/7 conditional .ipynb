
## If Statement
- If statement is used to execute a block of code only when a specified condition evaluates to True.


```python
age = 20
if age >= 18:
    print("Eligible to vote.")
```

```python
age = 19
if age > 18: print("Eligible to Vote.")
```

```python
age = 10
if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
```

## If-elif-else Statement
- elif statement is used to check multiple conditions in a program.
- It executes a block of code when its condition evaluates to True after previous conditions evaluate to False.

```python
                  # example 1
age = int (input("enter a age "))
if age < 13:
    print("child")
elif age < 19: 
    print("teenger")
elif age < 60:
    print("adult")
else:
    print("senior ")
```

## Nested if-else Statement
- Nested if-else statement is an if-else statement placed inside another if or else block. 
- It is used to check conditions within another condition.

```python
                 # example 1
day = input("enter day : ")
age = int(input("enter a age "))
price = 12
if day == "wenday":
    if age >= 18:
        price -= 2 
    else:
        price = 8
        price -= 2
    print("today is discount : ",price)
else :
    if age >= 18:
        price = 12
    else:
        price-= 4
    print ('today ticket price:', price)
    
    
```

## Conditional Expression (Ternary Operator)
Conditional Expression (Ternary Operator) is a short way to write an if-else statement in a single line. It is used when choosing between two values based on a condition.

```python
age = 20
s = "Adult" if age >= 18 else "Minor"
print(s)
```
