```python
         # break statement -> With the break statement we can stop the loop even if the while condition is true
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

```python
          # continues statement -> With the continue statement we can stop the current iteration, and continue with the next: 
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

```python
 # With the else statement we can run a block of code once when the condition no longer is true
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
```

```python
for x in range(6):
  if x == 3: break
  print(x)
else:
  print("Finally finished!")

#If the loop breaks, the else block is not executed.
```

```python
     # A nested loop is a loop inside a loop
      #The "inner loop" will be executed one time for each iteration of the "outer loop"
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```

```python
             # Count of Prime Numbers Between 1 and 10
count = 0
for n in range(1,11):
    
    pri = True 
    if n < 2:
        pri = False
    for i in range(2,n):
        if n%i == 0:
            pri = False
            break
    if pri:
        prim_no = n
        count +=1
    print(n,prim_no)
print(f"prime no 1 to 10 is =" ,count)   
    
```
