```python
students = ["modi", "trump"]
subjects = ["Math", "Science", "English"]

for student in students:
    index = 0
    while index < len(subjects):
        print(student, "-", subjects[index])
        index += 1
```

```python
                   # star triangle
a = int(input("enter the no "))
for i in range(1,a+1):
    for j in range(0,i):
        print("*",end=" ")
    print("")
```

```python
# addition of the no factorial 
# 1/1! + 2/2! + 3/3! + 4/4! = 

n = int(input('enter n'))

result = 0
fact = 1

for i in range(1,n+1):
  fact = fact * i
  result = result + i/fact

print(result)
```

```python
# find the prime no -> using break sys 
lo = int(input("enter lower range "))
upp = int(input("enter upper range "))
for i in range(lo , upp+1):
    for j in range(2,int(i**0.5)+1):     # 
        if i%j == 0:         # 11/2 -> 11/3.....
            break
    else :
        print(i)
```

```python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))

for i in range(lower,upper+1):
  for j in range(2,i):
    if i%j == 0:
      break
  else:
    print(i)
```

```python
n= 5
for i in range(1,n):
    for j in range(1,i+1):
        print(j,end="")
    for k in range(i-1,0,-1):
        print(k,end="")
    print()
```

```python
for i in range(2, 4):
    for j in range(1, 11):
        if i == j:
            break
        print(i, "*", j, "=", i * j)
    print()
```

```python
for i in range(2, 4):
    for j in range(1, 11):
        if i == j:
            continue
        print(i, "*", j, "=", i * j)
    print()
```
