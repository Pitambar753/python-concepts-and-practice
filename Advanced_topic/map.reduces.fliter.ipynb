# Python map() function

  - The map() function is used to apply a given function to every item of an iterable, such as a list or tuple, and returns a map object (which is an iterator)

SYNTAX->
map(function, iterable)
 - function: The function we want to apply to every element of the iterable.
 - iterable: The iterable whose elements we want to process.
 - We can also pass multiple iterables if our function accepts multiple arguments.


By default, the map() function returns a map object, which is an iterator. 
In many cases, we will need to convert this iterator to a list to work with the results directly
- Applies a function to every item in an iterable. Returns a map object (lazy iterator). Always wrap in list() to see output.
- map() is LAZY — it doesn't compute until you consume it (with list(), for loop, etc.)


```python
s = ['1', '2', '3', '4']
res = map(int, s)
print(list(res))
```

```python
nums = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, nums))
print(squared)
```

```python
# You can pass built-in functions directly: int, str, float, len — no lambda needed!

str_nums = ["10", "20", "30"]
int_nums = list(map(int, str_nums))
print(int_nums)
print(sum(int_nums))

pri = list(map(print,str_nums))  # print() ha built-in functions use kela ahe 
```

```python
a = [1, 2, 3, 4]

# Using custom function in "function" parameter
# This function is simply doubles the provided number
def double(val):
  return val*2

res = list(map(double, a))  # map is iterator
print(res)
```

###  map() with TWO iterables
- map() can take multiple iterables — stops at the shortest one.

```python

a = [1, 2, 3]
b = [10, 20, 30]
result = list(map(lambda x, y: x + y, a, b))
print(result)
```

```python
### map() on list of dicts
employees = [
  {"name": "Amit", "salary": 50000},
  {"name": "Sara", "salary": 60000},
]
# Give 10% hike
hiked = list(map(lambda e: {**e, "salary": e["salary"]*1.1}, employees))
print(hiked)
```

```python
words = [" hello ", " world ", " python "]
# Chain: strip then uppercase
result = list(map(str.upper, map(str.strip, words)))
print(result)
```

```python
a = [1, 2, 3, 4]

# Using lambda function in "function" parameter
# to double each number in the list
res = list(map(lambda x: x * 2, a))
print(res)
```

```python
a = [1, 2, 3]
b = [4, 5, 6]
res = map(lambda x, y: x + y, a, b)
print(list(res))
```

```python
fruits = ['apple', 'banana', 'cherry']
res = map(str.upper, fruits)
print(list(res))
```

```python
words = ['apple', 'banana', 'cherry']
res = map(lambda s: s[0], words)
print(list(res))
```

# reduce() in Python
 - The reduce(fun,seq) function is used to apply a particular function passed in its argument to all of the list elements mentioned in the sequence passed along. This function is defined in "functools" module

Syntax of reduce() ->  
functools.reduce(function, iterable[, initializer])
- from functools import reduce
- reduce(function, iterable, [initializer]) 

   - function: A function that takes two arguments and performs an operation on them.
   - iterable: An iterable whose elements are processed by the function.
   - initializer (optional): A starting value for the operation. If provided, it is placed before the first element in the iterable.
   - Reduces a list to a single value by cumulatively applying a function. Must import from functools. Think of it as "folding" a list.
  



```python
#### reduce works left to right: acc is accumulator (running result), x is next element.
```

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]
total = reduce(lambda acc, x: acc + x, nums)
print(total) # ((((1+2)+3)+4)+5) = 15
```

```python
### Find maximum using reduce
```

```python
from functools import reduce

nums = [3, 7, 1, 9, 2]
maximum = reduce(lambda a, b: a if a > b else b, nums)
print(maximum)
```

```python
from functools import reduce

# Function to add two numbers
def add(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
res = reduce(add, a)

print(res)  # Output: 15
```

```python
from functools import reduce

# Summing numbers with reduce and lambda
a = [1, 2, 3, 4, 5]
res = reduce(lambda x, y: x + y, a)

print(res)
```

```python
from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda a, b: a * b, nums, 1)
print(product)

# With empty list — initializer saves from error
safe = reduce(lambda a, b: a * b, [], 1)
print(safe)
```

```python
from functools import reduce

nested = [[1,2], [3,4], [5,6]]
flat = reduce(lambda acc, x: acc + x, nested)
print(flat)
```

```python
from functools import reduce

words = ["apple", "banana", "apple", "cherry", "banana", "apple"]

counts = reduce(lambda acc, w: {**acc, w: acc.get(w, 0) + 1}, words, {})
print(counts)
```

# filter() Function
- Filters elements from an iterable based on a function that returns True/False. 
Only items where function returns True are kept.
- filter(function, iterable) → filter object

```python
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
```

### Filter None / falsy values
- Pass None as function — filter keeps only truthy values. Removes 0, None, False, "", [].
- filter(None, list) is a classic CCEE tricky question! 0, False, None, "", [] are all filtered out.

```python
data = [1, None, "", 0, "hello", False, 42]
clean = list(filter(None, data))
print(clean)
```

### Filter strings by length


```python
names = ["AI", "Python", "C", "Django", "Go"]
long_names = list(filter(lambda n: len(n) > 3, names))
print(long_names)
```

### Filter passing students from dict list

```python
students = [
  {"name": "Raj", "marks": 55},
  {"name": "Mia", "marks": 32},
  {"name": "Dev", "marks": 78},
]
passed = list(filter(lambda s: s["marks"] >= 40, students))
print(passed)
print([s["name"] for s in passed])
```

### filter() + map() chained together
- Filter then transform — a very common real-world pattern tested in CCEE.

```python
nums = range(1, 11)
# Step 1: filter odds, Step 2: square them
result = list(map(
  lambda x: x**2,
  filter(lambda x: x % 2 != 0, nums)))
print(result)
```
