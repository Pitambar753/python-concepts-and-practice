# Python Lambda Functions
- Python Lambda Functions are anonymous functions means that the function is without a name.
-  As we already know the def keyword is used to define a normal function in Python. Similarly, the lambda keyword is used to define an anonymous function in Python\\\



## Syntax-> lambda arguments : expression 

- lambda: The keyword to define the function.
- arguments: A comma-separated list of input parameters (like in a regular function).
- expression: A single expression that is evaluated and returned

1. lambda Function ->
    - Definition -> Single expression with lambda
    - Name ->  Anonymous (or named if assigned).
    - Statements ->	Single expression only.
    - Documentation	-> Cannot have a docstring.
    - Reusability -> Best for short, temporary functions.
    - lambda function has no return value (infact , return a function)


2. Regular Function (def)
    -  Definition -> Multiple lines of code.
    -  Name ->  Must have a name.
    -  Statements -> Can include multiple statements
    -  Documentation	-> Can include docstrings.
    -  Reusability -> Better for reusable and complex logic.
   
#### 3. why use the lambda functions 
    - because they are used with HOF (higher order function)

```python
# # Using lambda
# sq = lambda x: x ** 2
# print(sq(3))

# # Using def
# def sqdef(x):
#     return x ** 2
# print(sqdef(3))
add = lambda x: x + 1
add(4)

```

### Lambda with conditional expression
- Use ternary (if-else in one line) inside lambda for simple decisions.
- Ternary inside lambda is a very common CCEE MCQ pattern. Remember: value_if_true if condition else value_if_false

```python
is_even = lambda n: "Even" if n % 2 == 0 else "Odd"
grade = lambda m: "Pass" if m >= 40 else "Fail"

print(is_even(7)) # Odd
print(grade(35)) # Fail
```


### Lambda inside sorted() — sort by key
- Lambda as the key= argument to sorted() is very commonly tested. Sort list of tuples or dicts by any field.

```python
students = [("Riya", 88), ("Aman", 72), ("Priya", 95)]

# Sort by marks (index 1)
result = sorted(students, key=lambda s: s[1])
print(result)

# Sort descending
result2 = sorted(students, key=lambda s: s[1], reverse=True)
print(result2)
```


### Immediately Invoked Lambda (IIFE)
- A lambda can be called immediately without storing it in a variable. This is a CCEE tricky output question.
- CCEE pattern: Wrap lambda in () then call with (). This trips up many students in output-prediction MCQs.

```python
result = (lambda x, y: x * y)(4, 5)
print(result)

# Power function immediately called
print((lambda b, e: b**e)(2, 10))
```


### Lambda returning lambda (closure)
- A lambda can return another lambda — creating a closure or function factory. Important for CDAC advanced Python.
- 

```python
# Multiplier factory
multiplier = lambda n: (lambda x: x ** n)

double = multiplier(2)
triple = multiplier(3)

print(double(4)) # 10
print(triple(3)) # 15
```

```python
# Example: Check if a number is positive, negative, or zero
n = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"

print(n(5))   
print(n(-3))  
print(n(0))
```

```python
li = [lambda arg=x: arg * 10 for x in range(1, 5)]
li2 = [lambda: x * 10 for x in range(1, 5)]
for i in li:
    print(i(),end=" - ")
print()
for i in li2:
    print(i(),end=" - ")

# lambda arg=x	Captures value immediately ✅
# lambda: x	Uses value later (late binding) ❌
```

```python
li23 = [lambda x=x: x * 10 for x in range(1, 5)]
for i in li23:
    print(i())
```

```python

# Example: Perform addition and multiplication in a single line
calc = lambda x, y: (x + y, x * y)

res = calc(3, 4)
print(type(calc(3, 4)))
print(res)
```

```python
# Example: Filter even numbers from a list
n = [1, 2, 3, 4, 5, 6]
even = filter(lambda x: x % 2 == 0, n)
print(list(even))
```

```python
# Example: Double each number in a list
a = [1, 2, 3, 4]
b = map(lambda x: x * 2, a)
print(list(b))
print(b)
```

```python
from functools import reduce

# Example: Find the product of all numbers in a list
a = [1, 2, 3, 4]
b = reduce(lambda x, y: x * y, a)
print(b)
```
