```python
# operator name is -> AND (&)

print(6 & 3)




#The & operator compares each bit and set it to 1 if both are 1, otherwise it is set to 0:
# 6 = 0000000000000110
# 3 = 0000000000000011
# --------------------
# 2 = 0000000000000010
# ====================

# Decimal numbers and their binary values:
# 0 = 0000000000000000
# 1 = 0000000000000001
# 2 = 0000000000000010
# 3 = 0000000000000011
# 4 = 0000000000000100
# 5 = 0000000000000101
# 6 = 0000000000000110
#7 = 0000000000000111 
```

```python
                 # operator name is -> OR (|) 

print(6 | 3)




# The | operator compares each bit and set it to 1 if one or both is 1, otherwise it is set to 0:

# 6 = 0000000000000110
# 3 = 0000000000000011
# --------------------
# 7 = 0000000000000111
# ====================

# Decimal numbers and their binary values:
# 0 = 0000000000000000
# 1 = 0000000000000001
# 2 = 0000000000000010
# 3 = 0000000000000011
# 4 = 0000000000000100
# 5 = 0000000000000101
# 6 = 0000000000000110
# 7 = 0000000000000111
```

```python
                 # operator name is -> XOR(^)

print(6 ^ 3)




# The ^ operator compares each bit and set it to 1 if only one is 1, otherwise (if both are 1 or both are 0) it is set to 0:

# 6 = 0000000000000110
# 3 = 0000000000000011
# --------------------
# 5 = 0000000000000101
# ====================

# Decimal numbers and their binary values:
# 0 = 0000000000000000
# 1 = 0000000000000001
# 2 = 0000000000000010
# 3 = 0000000000000011
# 4 = 0000000000000100
# 5 = 0000000000000101
# 6 = 0000000000000110
# 7 = 0000000000000111
```

```python
                # operator name is -> NOT (~)    

print(~3)




# The ~ operator inverts each bit (0 becomes 1 and 1 becomes 0).

# Inverted 3 becomes -4:
#  3 = 0000000000000011
# -4 = 1111111111111100

# Decimal numbers and their binary values:
#  4 = 0000000000000100
#  3 = 0000000000000011
#  2 = 0000000000000010
#  1 = 0000000000000001
#  0 = 0000000000000000
# -1 = 1111111111111111
# -2 = 1111111111111110
# -3 = 1111111111111101
# -4 = 1111111111111100
```

```python
                 # operator name is -> Zero fill left shift(<<)


print(3 << 2)




# The << operator inserts the specified number of 0's (in this case 2) from the right and let the same amount of leftmost bits fall off:

# If you push 00 in from the left:
#  3 = 0000000000000011
# becomes
# 12 = 0000000000001100

# Decimal numbers and their binary values:
#  0 = 0000000000000000
#  1 = 0000000000000001
#  2 = 0000000000000010
#  3 = 0000000000000011
#  4 = 0000000000000100
#  5 = 0000000000000101
#  6 = 0000000000000110
#  7 = 0000000000000111
#  8 = 0000000000001000
#  9 = 0000000000001001
# 10 = 0000000000001010
# 11 = 0000000000001011
# 12 = 0000000000001100
```

```python
            # operator name is -> Signed right shift(>>)

print(1 >> 2)




# The >> operator moves each bit the specified number of times to the right. Empty holes at the left are filled with 0's.

# If you move each bit 2 times to the right, 8 becomes 2:
#  8 = 0000000000001000
# becomes
#  2 = 0000000000000010

# Decimal numbers and their binary values:
#  0 = 0000000000000000
#  1 = 0000000000000001
#  2 = 0000000000000010
#  3 = 0000000000000011
#  4 = 0000000000000100
#  5 = 0000000000000101
#  6 = 0000000000000110
#  7 = 0000000000000111
#  8 = 0000000000001000
#  9 = 0000000000001001
# 10 = 0000000000001010
# 11 = 0000000000001011
# 12 = 0000000000001100
```

```python
# divmod ->Python is used to perform division and modulus operations simultaneously.
           # It returns a tuple containing two values: the quotient and the remainder
#divmod(a, b)
```

```python
result = divmod(10, 3)
print(result)  # Output: (3, 1)
# Explanation: 10 // 3 = 3 (quotient), 10 % 3 = 1 (remainder)
```

```python
result = divmod(-10, 3)
print(result)  # Output: (-4, 2)
# Explanation: -10 // 3 = -4 (quotient), -10 % 3 = 2 (remainder)
```

```python
result = divmod(10.5, 3)
print(result)  # Output: (3.0, 1.5)
# Explanation: 10.5 // 3 = 3.0 (quotient), 10.5 % 3 = 1.5 (remainder)
```

```python
# Example: Divide 25 by 4 repeatedly
dividend = 25
divisor = 4
while dividend >= divisor:
    quotient, remainder = divmod(dividend, divisor)
    print(f"Quotient: {quotient}, Remainder: {remainder}")
    dividend = remainder
```

```python
#id -> Python is used to get the unique identifier (memory address) of an object.
       #This identifier is guaranteed to be unique for the object during its lifetime
```

```python
x = 10
print(id(x))  # Output: A unique identifier (memory address), e.g., 140710469166928
# The id() function returns an integer that represents the memory address (or equivalent) where the object is stored
```

```python
x = 10
y = 10
print(id(x))  # Output: Unique ID for 10
print(id(y))  # Output: Same ID as x because integers are immutable and Python reuses objects

z = 20
print(id(z))  # Output: Different ID as z references a different value
```

```python
list1 = [1, 2, 3]
list2 = list1
print(id(list1))  # Output: Memory address of list1
print(id(list2))  # Output: Same as list1 because list2 references the same object

list1.append(4)
print(list2)  # Output: [1, 2, 3, 4] (changes are reflected in list2 as they share the same object)
```

```python
x = [1, 2, 3] 
y = [1, 2, 3]
print(id(x))  # Output: ID of x
print(id(y))  # Output: Different ID because x and y are different objects even if their content is identical
```

```python
#  NOTE
# 1) Immutable Objects: For immutable objects (like integers, strings, tuples),
     #Python may reuse the same object for optimization.

#  2)Mutable Objects: For mutable objects (like lists, dictionaries), 
      # different objects usually have different IDs, even if their content is identical.
```

```python
print(bin(10))  # Output: '0b1010' (binary representation of 10)
print(bin(255))  # Output: '0b11111111' (binary representation of 255)
print(bin(-10))  # Output: '-0b1010' (binary representation of -10)
    
```

```python
print(oct(10))  # Output: '0o12' (octal representation of 10)
print(oct(255))  # Output: '0o377' (octal representation of 255)
print(oct(-10))  # Output: '-0o12' (octal representation of -10)
```

```python
print(hex(10))  # Output: '0xa' (hexadecimal representation of 10)
print(hex(255))  # Output: '0xff' (hexadecimal representation of 255)
print(hex(-10))  # Output: '-0xa' (hexadecimal representation of -10)
```

```python
# ord ->Python is used to return the Unicode code point of a given character. 
       #It is the opposite of the chr() function, which converts a Unicode code point to a character
```

```python
print(ord('A'))  # Output: 65 (Unicode code point of 'A')
print(ord('a'))  # Output: 97 (Unicode code point of 'a')
print(ord('0'))  # Output: 48 (Unicode code point of '0')
print(ord('@'))  # Output: 64
print(ord('#'))  # Output: 35
print(ord('$'))  # Output: 36

```

```python
print(ord('₹'))  # Output: 8377 (Unicode code point of Indian Rupee symbol)
print(ord('π'))  # Output: 960 (Unicode code point of Greek letter Pi)
print(ord('🙂'))  # Output: 128578 (Unicode code point of the smiley emoji)
```

```python
# Print Unicode values of all characters in a string
for char in "Python":
    print(f"Character: {char}, Unicode: {ord(char)}")
# Output:
# Character: P, Unicode: 80
# Character: y, Unicode: 121
# Character: t, Unicode: 116
# Character: h, Unicode: 104
# Character: o, Unicode: 111
# Character: n, Unicode: 110
```

```python
# Printing emojis using their Unicode values
print("\U0001F600")  # 😀 Grinning Face
print("\U0001F602")  # 😂 Face with Tears of Joy
print("\U0001F970")  # 🥰 Smiling Face with Hearts
print("\u2764")      # ❤️ Red Heart
```

```python
# Generate a range of smiley emojis (U+1F600 to U+1F64F)
for code in range(0x1F600, 0x1F650):
    print(chr(code), end=" ")
```
