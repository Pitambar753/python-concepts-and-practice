@ Arithmetric Operators

```python

print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2)   # floor

print(5%2)  

print(5**2)
```

@ Relational Operators

```python
print(4>5)

print(4<5)

print(4>=5)

print(4<=5)

print(4==4)

print(4!=4)
```

@ 1.  Logical Operators

```python
print(1 and 0)

print(1 or 0)

print(not 1)
```

@ 1.Bitwise Operators

```python
# bitwise and
print(2 & 3)

# bitwise or
print(2 | 3)

# bitwise xor
print(2 ^ 3)

print(~3)

print(4 >> 2)

print(5 << 2)
```

@. Assignment Operators

```python
# = 
# a = 2
a = 2

# a = a % 2
a %= 2
a += 2
# a++ ++a

print(a)
```

@. Membership Operators

```python
# in/not in

print('D' not in 'Delhi')

print(1 in [2,3,4,5,6])
```

```python
number = int(input('Enter a 3 digit number'))

# 345%10 -> 5
a = number%10

number = number//10

# 34%10 -> 4
b = number % 10

number = number//10
# 3 % 10 -> 3
c = number % 10

print(a + b + c)
```
