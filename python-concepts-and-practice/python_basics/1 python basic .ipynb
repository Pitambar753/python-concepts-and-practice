### Key Idea
1. int(value, base) → always gives integer (decimal)
2. bin(), oct(), hex() → give string output

```python
# This converts a string from any base → decimal (int).
int('1010', 2)   # Binary → Decimal
print(int('12', 8))     # Octal → Decimal
int('64', 16)    # Hex → Decimal
int('100', 10)   # Decimal → Decimal

# int('hii',5)    # ValueError: invalid literal for int() with base 5: 'hii'
```

```python
int('100', 5)    #  1×5^2+ 0×5^1+0×5^0=25
int('100', 6)    # 
int('10', 6)
print(int('11',6))

type(int('1010', 2))  # <class 'int'>
```

```python
######### Python has built-in functions:
print(bin(10))   # Decimal → Binary
print(oct(10))   # Decimal → Octal
print(hex(10))   # Decimal → Hex

print(type(bin(10)))  # <class 'str'>
type(oct(10))
```

```python
####### sslicing method also you use it here  -> because it's string na 
print(bin(10)[2:])   # '1010'
oct(10)[2:]   # '12'
hex(10)[2:]   # 'a'
```
