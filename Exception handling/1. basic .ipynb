#### finally
- finally ha execute honar cha ahe , except jari call zala tari pan or else zari call zala tari pan execute honar cha brr ka
- ##### why gives finally
   - -> for code continute

```python
try :
    m = 0
    f = open("sample.txt",'r')
    print(m)
    print(5/2)
except FileNotFoundError:
    print("file not found")
except NameError:
    print("variable is not found")
except ZeroDivisionError:
    print("can't divide by zero ")
else:
    print(f.read())
finally:
    print("ye to print hoga hi")
```

```python
try :
    m = 0
    f = open("sample1.txt",'r')
    print(m)
    print(5/2)
except FileNotFoundError:
    print("file not found")
except NameError:
    print("variable is not found")
except ZeroDivisionError:
    print("can't divide by zero ")
else:
    print(f.read())
finally:
    print("ye to print hoga hi")
```

1. try block asks for user input and tries to divide 100 by the input number.
2. except blocks handle ZeroDivisionError and ValueError.
3. else block runs if no exception occurs, displaying the result.
4. finally block runs regardless of the outcome, indicating the completion of execution.

```python
try:
    n = 0
    res = 100 / n
    
except ZeroDivisionError:
    print("You can't divide by zero!")
    
except ValueError:
    print("Enter a valid number!")
    
else:
    print("Result is", res)
    
finally:
    print("Execution complete.")
```
