### `*args and **kwargs`
`*args and **kwargs` 

are special Python keywords that are used to pass the variable length of arguments to a function
- *args in Python (Non-Keyword Arguments)
- **kwargs in Python (Keyword Arguments)
- ##### Points to remember while using `*args and **kwargs`

- order of the arguments matter(normal -> `*args` -> `**kwargs`)
- The words “args” and “kwargs” are only a convention, you can use any name of your choice


```python
         # arbitrary argument 

def flexi(*p2):  # *arg name ahe na tuple chaya from madhe save hotaya 
    pro = 1
    print(p2)
    print(type(p2))
    for i in p2:
        pro = pro *i
        print(pro)
    print(pro)

    
flexi(1,2,3,5,6,8,10,12)
```

```python
def fun(arg1, *p2):
    print("First argument :", arg1)
    for arg in p2:
        print("Argument *argv :", arg)
    return type(p2)

fun('Hello', 'Welcome', 'to', 'My world')
```

```python
def fun(**kwargs):
    print(type(kwargs))
    for k, val in kwargs.items():
        print("%s == %s" % (val,k))


# Driver code
fun(s1='work', s2='form', s3='home')
```

```python
def fun(arg1, **kwargs):
    for k, val in kwargs.items():
        print("%s == %s" % (k, val))


# Driver code
fun("Hi", s1='I', s2='love', s3='you')
```

```python
def fun(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

fun(1, 2, 3, a=4, b=5)
```

# Passing function as an argument in Python

- first-class objects meaning they can be assigned to variables, 
passed as arguments and returned from other functions. 

This enables higher-order functions, decorators and lambda expressions. By passing a function as an argument, 


1. Higher Order Functions
   - A higher-order function takes or returns another function, enabling reusable and efficient code. It supports functional programming with features like callbacks, decorators, and utilities such as map(), filter(), and reduce().

```python
def process(func, text):  # applies a function to text  
    return func(text)  

def uppercase(text):  # converts text to uppercase  
    return text.upper()  

print(process(uppercase, "hello"))
```

```python
# higher-order function   # function la function retutn dene yala hof called 
def fun(func, number):
    return func(number)

# function to double a number
def double(x):
    return x * 2

print(fun(double, 5))
```

```python
# function to apply an operation on a list
def fun(func, numbers):
    return [func(num) for num in numbers]

# using the built-in 'abs' function
a = [-1, -2, 3, -4]
print(fun(abs, a))
```

```python
# function that applies an operation to a number
def fun(func, number):
    return func(number)

# passing a lambda function
print(fun(lambda x: x ** 2, 5))
```

# Returning a function from a function - Python

### Example 1: Immutable Data Types
  - with an integer, which is an immutable data type in Python:

```python
x = 10

def fun(x):
    x = 15

fun(x)
print(x)
```

```python
x = 10
def fun(x):
    print(id(x),'->ID of local x before reassignment')  	# ID of local x before reassignment
    x = 15
    print(id(x),'->ID of local x after reassignment')  	# ID of local x after reassignment


print(id(x),'-> ID of global x')  		# ID of global x
fun(x)
print(id(x),'-> ID of global x remains unchanged')  		# ID of global x remains unchanged
```

```python
def fun(l):
    l = [40, 50]  # Reassignment to a new list
    print(l)  
l = [10, 20, 30]
fun(l)
print(l)
```

### Example 2: Mutable Data Types
  - now, let's look at an example with a list, which is a mutable data type:

```python
l = [10, 20, 30]
def fun(l):
    l.append(15)

fun(l)
print(l)
```

```python
l = [10, 20, 30]

def fun(l):
    print(id(l))  # ID of local l before modification
    l.append(15)
    print(id(l))  # ID of local l after modification


print(id(l))  # ID of global l
fun(l)
print(id(l))  # ID of global l remains the same
```
