## any() function

1. any() checks if at least one value is True in an iterable (like list, tuple, etc.).
#### synatx 
- any(iterable)
1. note 
- Returns True → if any element is truthy
- Returns False → if all elements are false

```python
any([False, False, True])   # True
any([0, 0, 0])              # False
any([1, 0, 0])              # True
any([])                     # False
any(('',''))               # False
any(('p',''))               # true 
any('')
any({a:'a',b:'a'})
# any('','')                  # TypeError: any() takes exactly one argument (2 given)
```

```python
print(any([0, 0, 1, 0]))
print(any([False, "", None]))
print(any([]))
d = {'a': 0, 'b': 0, 'c': 1}
print(any(d))                # Remember: checks keys
d = {'a': 0, 'b': 0, 'c': 1}
print(any(d.values()))      # checks values only 
d = {0: 'x', 0: 'y'}
print(any(d))
print()                  
```

```python
d = {'a': 2, 'b': 4, 'c': 6}
a = [v % 2 == 1 for v in d.values()]
print(a)
any(v % 2 == 1 for v in d.values())
```

```python
d = {'a': 0, 'b': 0, 'c': 0}
any(v == 0 for v in d.values())
```

```python
d = {'a': 3, 'b': 5, 'c': 7}
any(v > 6 for v in d.values())
```

```python
print(any("hello"))
print(any([[], {}, 0, ""]))
print(any([[], {}, 0, "hi"]))
d = {'': 1}
print(any(d))
```

## Counter (from collections) function
1. Counter counts how many times each element appears.
   - syntax--->>(from collections import Counter)

```python
from collections import Counter

data = ['a', 'b', 'a', 'c', 'b', 'a']
c = Counter(data)

print(c)
print(c.most_common(1))   # Most frequent element
print(c.most_common(3))
c['a']             # Count of 'a'
```
