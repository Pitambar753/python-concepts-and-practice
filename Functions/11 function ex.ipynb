```python
                         # function (YIELD) -> is a generator
```

```python
def even_numbers(limit):
    for i in range(2, limit + 1, 2):  # Step by 2 to get even numbers
        yield i  # Yield the current even number, then pause

# Using the generator
for number in even_numbers(10):
    print(number)

```

```python
                      # yield
```

```python
def infinite_counter(start=0):
    while True:  # Infinite loop
        yield start  # Yield the current value
        start += 1   # Increment for the next iteration

# Using the generator
counter = infinite_counter()

# Fetch some values from the infinite generator
print(next(counter))  # Outputs: 0
print(next(counter))  # Outputs: 1
print(next(counter))  # Outputs: 2
# This can go on forever without exhausting memory
```

### Returning Multiple Values with a Tuple
#### How It Works
 - When the function returns the values, they are packed into a tuple. When these values are received, the tuple is unpacked into the variables 
 and  This is an efficient way to handle multiple return values.

```python
def calculate(a, b):
    sum_result = a + b
    mul_result = a * b
    return sum_result, mul_result


# Calling the function
s, m = calculate(10, 20)
print("Sum:", s)          
print("Multiplication:", m)
```

```python
def calculate_extended(a, b):
    sum_result = a + b
    mul_result = a * b
    sub_result = a - b
    return sum_result, mul_result, sub_result


# Calling the modified function
s, m, sub = calculate_extended(10, 20)
print("Sum:", s)         
print("Multiplication:", m) 
print("Subtraction:", sub)
```

```python
def calculate_list(a, b):
    sum_result = a + b
    mul_result = a * b
    sub_result = a - b
    return [sum_result, mul_result, sub_result]


# Calling the modified function
results = calculate_list(10, 20)
print("Results:", results)
```
