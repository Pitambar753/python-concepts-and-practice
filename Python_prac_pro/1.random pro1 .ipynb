```python
a = ['Learn', 'Quiz', 'Practice', 'Contribute'] 
b = a       
#The variable (b) references the same list as (a). Changes made to b will reflect in a because both reference the same object
c = a[:] 
#The variable (c) is a shallow copy of the list (a). Changes to c won't affect a or b since c refers to a new object.

b[0] = 'Code'   # This changes the first element of b to 'Code'. Since b and a are the same list, this change is reflected in a as well
c[1] = 'Mcq'    # This changes the second element of c to 'Mcq'. This change doesn't affect a or b because c is a copy

count = 0
for c in (a, b, c):   #loop iterates over a, b, and c
    if c[0] == 'Code': 
        count += 1
    if c[1] == 'Mcq': 
        count += 10

print(count)
```

## prime factors

```python
def prime_factors(n):
    # Divide by 2 until n becomes odd
    while n % 2 == 0:
        print(2, end=' ')
        n = n // 2

    # Now n must be odd. Start from 3 and go up to sqrt(n)
    i = 3
    while i * i <= n:
        while n % i == 0:
            print(i, end=' ')
            n = n // i
        i += 2

    # If n is still greater than 2, it must be a prime number
    if n > 2:
        print(n)

# Example Usage
n = int(input("Enter a number: "))
prime_factors(n)
```
