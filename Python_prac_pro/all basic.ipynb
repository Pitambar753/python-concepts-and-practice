#### Decimal number to binary number

- Given a decimal number n (positive) in string format, compute its binary string equivalent and return it. 
- Note: Don't add a new line at the end.

 1. Examples:

 - Input: n = 7
 - Output: 111

```python
print(1//2)
print(1%2)
```

```python
def decimal_to_binary(n):
    if n == 0:
        return "0"
    
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary
        n = n // 2
    return binary


# Example usage
number = 12
print("Binary equivalent:", decimal_to_binary(number))
```

```python
def decimal_to_binary(n):
    # Convert the string to integer
    num = int(n)
    
    # Convert integer to binary using bin(), remove '0b' prefix
    binary = bin(num)[2:]
    
    return binary

    

decimal_to_binary(7)  
```

#### Binary number to decimal number

- Given a string b representing a Binary Number, The problem is to find its decimal equivalent.
1. Examples:

- Input : b = 111
- Output : 7
- Explanation : The decimal equivalent of the binary number 111 is 22 + 21 + 20 = 7.
2.
 - Input : b = 1010
 - Output : 10
 - Explanation : The decimal equivalent of the binary number 1010 is 23 + 21 = 10.

```python
def binary_to_decimal(binary_str):
    result = 0
    power = 1

    # Traverse the string from the end to the beginning
    for bit in reversed(binary_str):
        result += int(bit) * power
        power *= 2
    
    return result

# Example usage
binary_input = "110"
print("Decimal equivalent:", binary_to_decimal(binary_input))
```

```python
def binary_to_decimal_builtin(binary_str):
    return int(binary_str, 2)

# Example usage
binary_input = "110"
print("Decimal equivalent:", binary_to_decimal_builtin(binary_input))
```

### Average
- You are given a list arr that contains integers. You need to return average of the non negative integers.
1. Examples:

 - Input: arr = [-12, 8, -7, 6, 12, -9, 14]
 - Output: avg = 10.0
 - Explanation: The non negative numbers are 8 6 12 14. The sum is 8+6+12+14 = 40, Average = 40/4 = 10.0
2. ex:
- Input: arr = [1, 2, 3]
- Output: avg = 2.0
- Explanation: The non negative numbers are 1 2 3. The sum is 1+2+3 = 6, Average = 6/3 = 2.0

```python
def average_non_negative(arr):
    # Filter non-negative numbers
    non_negatives = [i for i in arr if i >= 0]
    
    # Avoid division by zero
    if not non_negatives:
        return 0.0  # or you can return 'None' if preferred
    
    # Calculate average
    avg = sum(non_negatives) / len(non_negatives)
    return avg
arr = [5, 0, 0, 0]
average_non_negative(arr)
```

- Given a list of numbers, the goal is to determine if the list is sorted in non-decreasing order. A list is considered sorted if every element is either greater than or equal to the previous element.

1. Example 1:
- List: [10, 20, 30]
- Output: True (The list is sorted)

2. Example 2:
- List: [10, 20, 5, 30]
- Output: False (The list is not sorted because 5 is less than 20)

```python
def is_sorted(input_list):
    for i in range(1, len(input_list)):
        # Compare current element with the previous one
        if input_list[i] < input_list[i - 1]:
            return False
    return True
print(is_sorted([10, 20, 30]))  
print(is_sorted([10, 20, 5, 30]))
```

```python
def is_sorted(input_list):
    return input_list == sorted(input_list)
print(is_sorted([10, 20, 30]))  
print(is_sorted([10, 20, 5, 30]))
```

```python
def count_distinct_elements(input_list):
    count = 1
    for i in range(1, len(input_list)):
        if input_list[i] not in input_list[:i]:
            count += 1
    return count

numbers = [10, 20, 10, 30, 30, 20]
print(count_distinct_elements(numbers))
```

```python
class Complex:

    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def print(self):
        print(str(self.real) + "+" + str(self.imag) + "i")

    def add(self, C):
        self.real += C.real
        self.imag += C.imag

C1 = Complex(10, 20)
C1.print()  # Output: 10+20i

C2 = Complex(20, 30)
C1.add(C2)
C1.print()  # Output: 30+50i
```
