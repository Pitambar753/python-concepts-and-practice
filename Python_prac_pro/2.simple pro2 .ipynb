### GCD
  -  Given two numbers a and b. The task is to find the GCD of  a and b.
The GCD of two numbers is the largest number that can divide both a and b perfectly
1. Input: a = 6, b = 9
- Output: 3
- Explanation: After 3 there is no number that can divide both 6 and 9 perfectly.

```python
a = int(input())
b = int(input())
gcd = 1
for i in range(1,min(a,b)+1):
    if a%i == 0 and b%i == 0:
        gcd = i
print(gcd)
```

### calculator problem


```python
import sys

print("""
Please select an operation:
1. Addition
2. Subtraction
3. Multiplication
""")

# Read the user's choice
choice = int(input("Select operation from 1, 2, or 3: "))

# Check if the choice is valid
if choice not in (1, 2, 3):
    print("Invalid choice")
    sys.exit()

# Prompt the user to enter two numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform the selected operation
if choice == 1:
    result = num1 + num2
elif choice == 2:
    result = num1 - num2
else:
    result = num1 * num2

# Print the result
print("The result is: ", result)
```
